```python
import pandas as pd
import numpy as np

nykaa_df = pd.read_csv('/content/nykaa_campaign_data_with_nulls.csv')
purplle_df = pd.read_csv('/content/purplle_campaign_data_with_nulls.csv')
tira_df = pd.read_csv('/content/tira_campaign_data_with_nulls.csv')
```

```python
nykaa_df['Brand'] = 'Nykaa'
purplle_df['Brand'] = 'Purplle'
tira_df['Brand'] = 'Tira'

df=pd.concat([nykaa_df, purplle_df, tira_df], ignore_index=True)
```

```python
df.drop_duplicates(inplace=True)

df['Revenue'] = df['Revenue'].fillna(0)

ac_mean = df['Acquisition_Cost'].mean()
df['Acquisition_Cost'] = df['Acquisition_Cost'].fillna(ac_mean)

numeric_cols = ['Duration', 'Impressions', 'Clicks', 'Leads', 'Conversions', 'Engagement_Score']
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
  mode_series = df[col].mode()
  fallback_value = mode_series.iloc[0] if not mode_series.empty else 'Unknown'
  df[col] = df[col].fillna(fallback_value)

df['Duration'] = df['Duration'].astype(int)
df['Impressions'] = df['Impressions'].astype(int)
df['Clicks'] = df['Clicks'].astype(int)
df['Leads'] = df['Leads'].astype(int)
df['Conversions'] = df['Conversions'].astype(int)
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y', errors='coerce')
```

```python
df['ROI'] =(df['Revenue'] - df['Acquisition_Cost']) / df['Acquisition_Cost']

df['ROI'] = df['ROI'].round(2)
```

```python
df
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

df.columns = df.columns.str.lower()

if 'roi' in df.columns:
    df['profit_loss_flag'] = np.where(
        df['roi'] > 0, 'Profit', 'Loss'
    )

if 'channel_used' in df.columns:
    channels_split = (
        df['channel_used']
        .astype(str)
        .str.lower()
        .str.replace(' ', '')
        .str.split(',')
    )

    mlb = MultiLabelBinarizer()
    channel_encoded = mlb.fit_transform(channels_split)

    channel_df = pd.DataFrame(
        channel_encoded,
        columns=[f'channel_{cls}' for cls in mlb.classes_],
        index=df.index
    )

    df = pd.concat([df, channel_df], axis=1)

columns_to_drop = [
    'campaign_id',
    'date',
    'channel_used'
]
df = df.drop(columns=columns_to_drop, errors='ignore')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

cat_selected_col = [ 'brand']
Cat_class = []
for col in cat_selected_col:
    ordered_categories = df.groupby(col)['revenue'].mean().sort_values(ascending=True).index.tolist()
    Cat_class.append(ordered_categories)
encoder = OrdinalEncoder(categories=Cat_class)
df1[cat_selected_col] = encoder.fit_transform(df1[cat_selected_col])
numeric_cols = ['impressions', 'clicks', 'leads', 'conversions', 'acquisition_cost','engagement_score', 'channel_email', 'channel_facebook',
       'channel_google', 'channel_instagram', 'channel_whatsapp',
       'channel_youtube','roi']
features = numeric_cols + cat_selected_col

X = df1[features]
y = df1['revenue']
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import joblib
import plotly.express as px
import plotly.graph_objects as go

st.set_page_config(
    page_title="Marketing Analytics Dashboard",
    layout="wide",
    initial_sidebar_state="expanded"
)

@st.cache_resource
def load_models():
    revenue_model = joblib.load("revenue_model.pkl")
    profit_model = joblib.load("profit_loss_model.pkl")
    return revenue_model, profit_model

try:
    revenue_model, profit_model = load_models()
except Exception as e:
    st.error(f"Error loading models: {e}. Make sure the .pkl files are in the same folder.")
    st.stop()

st.title("Campaign Performance Prediction Dashboard")
st.markdown("Provide your core metrics below to generate machine learning revenue predictions.")

st.sidebar.header("Campaign Setup")

brand_selection = st.sidebar.selectbox("Select Brand", ["Nykaa", "Purplle", "Tira"])

st.sidebar.markdown("**Select Channels Used:**")
use_email = st.sidebar.checkbox("Email Channel", value=True)
use_facebook = st.sidebar.checkbox("Facebook Channel", value=False)
use_google = st.sidebar.checkbox("Google Channel", value=False)
use_instagram = st.sidebar.checkbox("Instagram Channel", value=False)
use_whatsapp = st.sidebar.checkbox("WhatsApp Channel", value=False)
use_youtube = st.sidebar.checkbox("YouTube Channel", value=False)

st.sidebar.divider()
st.sidebar.header(" Primary Performance Metrics")

impressions = st.sidebar.number_input("Impressions", min_value=0, value=50000, step=1000)

clicks = st.sidebar.number_input("Clicks", min_value=0, value=2500, step=100)

acquisition_cost = st.sidebar.number_input("Campaign Spend / Cost (₹)", min_value=0.0, value=25000.0, step=500.0)

predict = st.sidebar.button("Run Predictions", use_container_width=True)

if predict:
    brand_mapping = {"Nykaa": 0, "Purplle": 1, "Tira": 2}

    DEFAULT_ENGAGEMENT_SCORE = 15.0

    estimated_leads = int(clicks * 0.20)
    estimated_conversions = int(estimated_leads * 0.20)

    raw_input = {
        "impressions": impressions,
        "clicks": clicks,
        "leads": estimated_leads,
        "conversions": estimated_conversions,
        "acquisition_cost": acquisition_cost,
        "engagement_score": DEFAULT_ENGAGEMENT_SCORE,
        "channel_email": 1 if use_email else 0,
        "channel_facebook": 1 if use_facebook else 0,
        "channel_google": 1 if use_google else 0,
        "channel_instagram": 1 if use_instagram else 0,
        "channel_whatsapp": 1 if use_whatsapp else 0,
        "channel_youtube": 1 if use_youtube else 0,
        "roi": 0.0,
        "brand": brand_mapping[brand_selection]
    }

    expected_feature_order = [
        "impressions", "clicks", "leads", "conversions",
        "acquisition_cost", "engagement_score", "channel_email",
        "channel_facebook", "channel_google", "channel_instagram",
        "channel_whatsapp", "channel_youtube", "roi", "brand"
    ]

    temp_df = pd.DataFrame([raw_input])[expected_feature_order]

    try:
        revenue_prediction = float(revenue_model.predict(temp_df)[0])

        if revenue_prediction < acquisition_cost:
            dynamic_roi = -1.0
        else:
            dynamic_roi = ((revenue_prediction - acquisition_cost) / acquisition_cost) if acquisition_cost > 0 else 0.0

        final_input = raw_input.copy()
        final_input["roi"] = dynamic_roi

        final_df = pd.DataFrame([final_input])[expected_feature_order]

        profit_prediction = profit_model.predict(final_df)[0]

    except ValueError as structural_err:
        st.error(f"Feature sequence check failed: {structural_err}")
        st.stop()

    st.subheader("Campaign Input Summary")
    col1, col2, col3 = st.columns(3)
    col1.metric("Selected Brand", brand_selection)
    col2.metric("Impressions Provided", f"{impressions:,}")
    col3.metric("Clicks Provided", f"{clicks:,}")

    st.divider()

    st.subheader("Prediction Results")
    r1, r2 = st.columns(2)
    r1.metric("Predicted Revenue", f"₹{revenue_prediction:,.2f}")

    with r2:
        st.write("**Profitability Status**")
        if profit_prediction in [1, "Profit", "PROFIT", 1.0]:
            st.success("PROFIT")
        else:
            st.error("LOSS")

    st.divider()

    graph_col1, graph_col2 = st.columns(2)
    with graph_col1:
        st.subheader("Funnel Estimate Analysis")
        funnel = go.Figure(go.Funnel(
            y=["Impressions", "Clicks", "Estimated Leads", "Estimated Conversions"],
            x=[impressions, clicks, estimated_leads, estimated_conversions],
            textinfo="value+percent initial"
        ))
        funnel.update_layout(margin=dict(l=20, r=20, t=30, b=20))
        st.plotly_chart(funnel, use_container_width=True)

    with graph_col2:
        st.subheader(" Financial Breakdown")
        compare_df = pd.DataFrame({
            "Category": ["Campaign Cost", "Predicted Revenue"],
            "Amount": [acquisition_cost, revenue_prediction]
        })
        fig2 = px.bar(compare_df, x="Category", y="Amount", text="Amount", color="Category")
        fig2.update_traces(texttemplate="₹%{text:,.2f}", textposition="outside")
        st.plotly_chart(fig2, use_container_width=True)

else:
    st.info("click **'Run Predictions'**.")
```
