```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df = df.drop_duplicates()
```

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
for col in numeric_cols:
  df[col] = df[col].fillna(df[col].median())
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
  mode_series = df[col].mode()
  fallback_value = mode_series.iloc[0] if not mode_series.empty else 'Unknown'
  df[col] = df[col].fillna(fallback_value)
```

```python
for col in categorical_cols:
  if 'id' in col:
    df[col] = df[col].astype(str).str.strip().str.upper()
  else:
    df[col] = df[col].astype(str).str.strip().str.title()

for col in df.columns:
  if "date" in col:
    df[col] = pd.to_datetime(df[col], format='%d-%m-%Y', errors='coerce')
```

```python
if 'roi' in df.columns and 'revenue' in df.columns and 'acquisition_cost' in df.columns:
  df['roi'] = np.where(
  df['acquisition_cost'] > 0,
  (df['revenue'] - df['acquisition_cost']) / df['acquisition_cost'],0)
  df['roi'] = df['roi'].round(2)
elif 'roi' in df.columns:
  df['roi'] = df['roi'].apply(lambda x: np.nan if x < 0 else x)
  df['roi'] = df['roi'].fillna(df['roi'].median())
```

```python
from scipy.stats import pearsonr,ttest_ind,chi2_contingency,f_oneway
```

```python
def Person_corr(df,f_col,t_col):
    corr,p_val=pearsonr(df[f_col],df[t_col])

    if p_val<0.01:
        return {'Feature':f_col,'Target':t_col,'Correlation':corr,'p-value':p_val,'Significant':True}
    else:
        return {'Feature':f_col,'Target':t_col,'Correlation':corr,'p-value':p_val,'Significant':False}
```

```python
def ANOVA_test(df,f_col,t_col):
    group=[df[df[f_col]==cls][t_col] for cls in df[f_col].unique()]
    f_stat, p_val = f_oneway(*group) # ANOVA Test

    aplha=0.05
    if p_val< aplha:
        return {'Feature':f_col,'Target':t_col,'F-statistic':f_stat,'p-value':p_val,'Significant':True}
    else:
        return {'Feature':f_col,'Target':t_col,'F-statistic':f_stat,'p-value':p_val,'Significant':False}
```

```python
def two_samp_test(df,Cat_col,Num_col):
    group=[df[df[Cat_col]==cls][Num_col] for cls in df[Cat_col].unique()]
    t_stat, p_val = ttest_ind(*group) # Two Sample Test

    aplha=0.05
    if p_val< aplha:
        return {'Feature':Cat_col,'Target':Num_col,'T-statistic':t_stat,'p-value':p_val,'Significant':True}
    else:
        return {'Feature':Cat_col,'Target':Num_col,'T-statistic':t_stat,'p-value':p_val,'Significant':False}
```

```python
def Chi_SQR_test(df,f_col,t_col):
    contingency_table=pd.crosstab(df[f_col],df[t_col])
    chi2, p_val, dof, expected = chi2_contingency(contingency_table)

    if p_val<0.01:
        return {'Feature':f_col,'Target':t_col,'Chi2-statistic':chi2,'p-value':p_val,'Significant':True}
    else:
        return {'Feature':f_col,'Target':t_col,'Chi2-statistic':chi2,'p-value':p_val,'Significant':False}
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler, StandardScaler, OrdinalEncoder, PolynomialFeatures
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, classification_report, accuracy_score
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier


revenue_features = ['clicks', 'leads', 'conversions', 'acquisition_cost', 'impressions', 'engagement_score']
X_reg = df[revenue_features]
y_reg = df['revenue']

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)


classification_features = ['clicks', 'leads', 'conversions', 'acquisition_cost', 'impressions', 'engagement_score']
X_cls = df[classification_features]


y_cls = df['profit_loss_flag'].map({'Profit': 1, 'Loss': 0})


X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(X_cls, y_cls, test_size=0.2, random_state=42)

print("Data splits ready!")
print(f"Regression train shape: {X_train_reg.shape}, Test shape: {X_test_reg.shape}")
print(f"Classification train shape: {X_train_cls.shape}, Test shape: {X_test_cls.shape}")
```

```python

REVENUE_Num_Selected = ['clicks', 'leads', 'conversions', 'acquisition_cost', 'impressions']
REVENUE_Dis_Selected = ['engagement_score']
REVENUE_Cat_Selected = []


train_temp = pd.concat([X_train_reg, y_train_reg], axis=1)
Cat_class = []
for col in REVENUE_Cat_Selected:
    if col in train_temp.columns:
        ordered_idx = train_temp.groupby(col)['revenue'].mean().sort_values(ascending=False).index.tolist()
        Cat_class.append(ordered_idx)

Num_Cont_transformer = Pipeline(steps=[
    ('Imputer', SimpleImputer(strategy='median')),
    ('Scaler', RobustScaler())
])

Num_Disc_Transformer = Pipeline(steps=[
    ('Imputer', SimpleImputer(strategy='most_frequent'))
])

CAT_Feat_Transformer = Pipeline(steps=[
    ('Imputer', SimpleImputer(strategy='most_frequent')),
    ('Encoder', OrdinalEncoder(categories=Cat_class))
])

Preprocess_Step = ColumnTransformer(transformers=[
    ('Num_Cont_Transformer', Num_Cont_transformer, REVENUE_Num_Selected),
    ('Num_Disc_Transformer', Num_Disc_Transformer, REVENUE_Dis_Selected),
    ('CAT_Feat_Transformer', CAT_Feat_Transformer, REVENUE_Cat_Selected)
])


def evaluate_regression_model(model_name, pipeline_object):
    pipeline_object.fit(X_train_reg, y_train_reg)
    y_tr_pred = pipeline_object.predict(X_train_reg)
    y_te_pred = pipeline_object.predict(X_test_reg)

    print(f"=== {model_name} Results ===")
    print(f"Train MAE: {mean_absolute_error(y_train_reg, y_tr_pred):.2f} | Test MAE: {mean_absolute_error(y_test_reg, y_te_pred):.2f}")
    print(f"Train R² Score: {r2_score(y_train_reg, y_tr_pred):.4f} | Test R² Score: {r2_score(y_test_reg, y_te_pred):.4f}\n")


lr_pipeline = Pipeline([('Preprocess', Preprocess_Step), ('Model', LinearRegression())])
evaluate_regression_model("Linear Regression", lr_pipeline)

poly_pipeline = Pipeline([('Preprocess', Preprocess_Step), ('Poly', PolynomialFeatures(degree=2)), ('Model', LinearRegression())])
evaluate_regression_model("Polynomial Regression (Degree 2)", poly_pipeline)


dt_reg_pipeline = Pipeline([('Preprocess', Preprocess_Step), ('Model', DecisionTreeRegressor(max_depth=5, min_samples_split=10, random_state=42))])
evaluate_regression_model("Decision Tree Regressor", dt_reg_pipeline)

rf_reg_pipeline = Pipeline([('Preprocess', Preprocess_Step), ('Model', RandomForestRegressor(n_estimators=100, max_depth=10, min_samples_split=5, random_state=42, n_jobs=-1))])
evaluate_regression_model("Random Forest Regressor", rf_reg_pipeline)
```

```python

cls_preprocessor = ColumnTransformer(transformers=[
    ('Num_Transformer', Pipeline(steps=[
        ('Imputer', SimpleImputer(strategy='median')),
        ('Scaler', StandardScaler())
    ]), classification_features)
])


def evaluate_classification_model(model_name, pipeline_object):
    pipeline_object.fit(X_train_cls, y_train_cls)
    y_pred = pipeline_object.predict(X_test_cls)

    print(f"==================================================")
    print(f" {model_name.upper()} PERFORMANCE REPORT")
    print(f"==================================================")
    print(f"Accuracy Score: {accuracy_score(y_test_cls, y_pred):.4f}\n")
    print(classification_report(y_test_cls, y_pred, target_names=['Loss', 'Profit']))
    print("\n")


log_reg_pipeline = Pipeline([('Preprocessor', cls_preprocessor), ('Classifier', LogisticRegression(class_weight='balanced',random_state=42))])
evaluate_classification_model("Logistic Regression", log_reg_pipeline)


dt_cls_pipeline = Pipeline([('Preprocessor', cls_preprocessor), ('Classifier', DecisionTreeClassifier(max_depth=6, min_samples_leaf=4, random_state=42))])
evaluate_classification_model("Decision Tree Classifier", dt_cls_pipeline)

rf_cls_pipeline = Pipeline([('Preprocessor', cls_preprocessor), ('Classifier', RandomForestClassifier(n_estimators=150, max_depth=12, min_samples_split=4, random_state=42, n_jobs=-1))])
evaluate_classification_model("Random Forest Classifier", rf_cls_pipeline)
```
